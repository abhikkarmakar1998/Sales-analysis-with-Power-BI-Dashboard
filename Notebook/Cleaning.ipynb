#### Importing Library

```python
import pandas as pd

df = pd.read_excel('../Data/sales_raw_500.xlsx')
```

```python
df
```

#### Remove Duplicates

```python
df = df.drop_duplicates()
```

```python
df
```

#### Fix Date Column

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
```

#### Ordering Date

```python
df = df.dropna(subset=['Order_Date'])
```

```python
df
```

#### Generating Sales, Month & Year column

```python
df['Profit'] = df['Sales'] - df['Cost']
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
```

```python
df
```

#### Cleaned Dataset

```python
df.to_csv('Sales_Cleaned.csv', index= False)
```
